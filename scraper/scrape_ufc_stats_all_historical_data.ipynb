```python
import time
# --- create empty dfs just for display (optional)
all_fight_results_df = pd.DataFrame(columns=config['fight_results_column_names'])
all_fight_stats_df   = pd.DataFrame(columns=config['fight_stats_column_names'])

ckpt_path = Path("processed_fight_urls.ckpt")

# Load checkpoint
done = set()
if ckpt_path.exists():
    done = set(line.strip() for line in ckpt_path.read_text().splitlines() if line.strip())

# Filter out already processed
todo = [u for u in list_of_fight_details_urls if u not in done]

# For a 10-URL test, uncomment:
# todo = todo[:10]

print(f"To process now: {len(todo)} | Already done: {len(done)}")

def append_csv(df: pd.DataFrame, path: str, columns: list):
    """Append rows to CSV with consistent schema; write header only if new."""
    if df is None or df.empty:
        return
    p = Path(path)
    p.parent.mkdir(parents=True, exist_ok=True)
    header = not p.exists() or p.stat().st_size == 0
    df.reindex(columns=columns).to_csv(p, mode="a", header=header, index=False)

fail_log = []

# loop through each fight and parse fight results and stats
for url in tqdm_notebook(todo):
    try:
        # get soup
        soup = LIB.get_soup(url)

        # parse fight results and fight stats
        fight_results_df, fight_stats_df = LIB.parse_organise_fight_results_and_stats(
            soup,
            url,
            config['fight_results_column_names'],
            config['totals_column_names'],
            config['significant_strikes_column_names']
        )

        # append to disk immediately (so you never lose more than one URL)
        append_csv(fight_results_df, config['fight_results_file_name'], config['fight_results_column_names'])
        append_csv(fight_stats_df,   config['fight_stats_file_name'],   config['fight_stats_column_names'])

        # also keep small in-memory copies just for display (optional)
        all_fight_results_df = pd.concat([all_fight_results_df, fight_results_df], ignore_index=True)
        all_fight_stats_df   = pd.concat([all_fight_stats_df,   fight_stats_df],   ignore_index=True)

        # mark checkpoint only after successful writes
        with open(ckpt_path, "a") as f:
            f.write(url + "\n")

        # polite delay to avoid hammering the site
        time.sleep(0.4)

    except Exception as e:
        print(f"⚠️ Failed: {url} -> {e}")
        fail_log.append({"URL": url, "ERROR": str(e)})
        time.sleep(1.0)
        continue

# show samples (avoid rendering thousands of rows)
display(all_fight_results_df.head(10))
display(all_fight_stats_df.head(10))

print(f"Done. Success: {len(todo) - len(fail_log)} | Failed: {len(fail_log)}")

# OPTIONAL: persist a fail log
if fail_log:
    fail_df = pd.DataFrame(fail_log)
    fail_path = Path(config["fight_stats_file_name"]).with_name("failed_fights.csv")
    if fail_path.exists():
        prev = pd.read_csv(fail_path)
        fail_df = pd.concat([prev, fail_df], ignore_index=True).drop_duplicates(subset=["URL"], keep="last")
    fail_df.to_csv(fail_path, index=False)
    print(f"⚠️ Fail log written -> {fail_path}")
```

```python
from pathlib import Path
from IPython.display import display

results_path = Path(config['fight_results_file_name'])
stats_path   = Path(config['fight_stats_file_name'])

def norm_text(s: pd.Series) -> pd.Series:
    s = s.astype(str).fillna("")
    s = s.str.replace(r"\s+", " ", regex=True).str.strip()
    s = s.str.replace(r"\s+vs\.\s+", " vs. ", regex=True)
    return s

def id_from_url(s: pd.Series) -> pd.Series:
    return s.astype(str).str.strip().str.rstrip("/").str.split("/").str[-1]

def normalize_fight_id_col(df: pd.DataFrame) -> pd.DataFrame:
    # Rename any case variant (fight_id, Fight_id, etc.) to FIGHT_ID
    for c in list(df.columns):
        if c.lower() == "fight_id" and c != "FIGHT_ID":
            df = df.rename(columns={c: "FIGHT_ID"})
            break
    return df

# --- 1) RESULTS: ensure FIGHT_ID from URL (read as str to avoid dtype warnings)
res = pd.read_csv(results_path, dtype=str)
res = normalize_fight_id_col(res)
res["FIGHT_ID"] = id_from_url(res["URL"])
res = res[["FIGHT_ID"] + [c for c in res.columns if c != "FIGHT_ID"]]
res.to_csv(results_path, index=False)
print(f"✅ Updated results with FIGHT_ID -> {results_path} (rows: {len(res)})")

# Build mapping (EVENT,BOUT) -> FIGHT_ID
map_df = res[["EVENT", "BOUT", "FIGHT_ID"]].copy()
map_df["EVENT_JOIN"] = norm_text(map_df["EVENT"])
map_df["BOUT_JOIN"]  = norm_text(map_df["BOUT"])
map_df = map_df.drop_duplicates(subset=["EVENT_JOIN", "BOUT_JOIN"])

# --- 2) STATS: add/overwrite FIGHT_ID via join
st = pd.read_csv(stats_path, dtype=str)
st = normalize_fight_id_col(st)

st["EVENT_JOIN"] = norm_text(st["EVENT"])
st["BOUT_JOIN"]  = norm_text(st["BOUT"])

# Merge mapping as FIGHT_ID_MAP to avoid suffix chaos
right = map_df[["EVENT_JOIN", "BOUT_JOIN", "FIGHT_ID"]].rename(columns={"FIGHT_ID": "FIGHT_ID_MAP"})
st = st.merge(right, on=["EVENT_JOIN", "BOUT_JOIN"], how="left", validate="many_to_one")

# Coalesce: keep existing FIGHT_ID if present, else take mapped one
if "FIGHT_ID" in st.columns:
    st["FIGHT_ID"] = st["FIGHT_ID"].where(st["FIGHT_ID"].notna() & (st["FIGHT_ID"].str.strip() != ""), st["FIGHT_ID_MAP"])
else:
    st["FIGHT_ID"] = st["FIGHT_ID_MAP"]

# Report misses (no id found from mapping)
missing = int(st["FIGHT_ID"].isna().sum())
if missing:
    print(f"⚠️ Could not assign FIGHT_ID to {missing} stat rows. A few unmatched (EVENT,BOUT):")
    display(st.loc[st["FIGHT_ID"].isna(), ["EVENT","BOUT"]].drop_duplicates().head(10))

# Reorder and drop helpers
st = st[["FIGHT_ID"] + [c for c in st.columns if c not in {"FIGHT_ID","EVENT_JOIN","BOUT_JOIN","FIGHT_ID_MAP"}]]

st.to_csv(stats_path, index=False)
print(f"✅ Updated stats with FIGHT_ID -> {stats_path} (rows: {len(st)})")
```
